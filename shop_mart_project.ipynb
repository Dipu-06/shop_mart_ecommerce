```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import f1_score, classification_report, confusion_matrix
```

```python
x=df.drop("Revenue",axis=1)
y=df["Revenue"].astype(int)
```

```python
cat_values=["Month","VisitorType"]
num_val=x.select_dtypes(include=["int64","float64"]).columns
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
transformer=ColumnTransformer(
    transformers=[
        ('scaler',StandardScaler(),num_val),
    
        ('ohn',OneHotEncoder(handle_unknown="ignore"),cat_values)
    ]
)
```

```python
dt=DecisionTreeClassifier(
     max_depth=6,              # prevents deep overfitting
    min_samples_leaf=30,      # smooths decision boundaries
    class_weight="balanced",  # handles imbalance
    random_state=42
)
pipe=Pipeline(
    steps=[
        ("transformer",transformer),
        ("tree",dt)
    ]
)
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pred = pipe.predict(x_test)

print("F1 score:", f1_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```
